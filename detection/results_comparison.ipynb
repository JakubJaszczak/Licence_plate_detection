```python
import json
from pathlib import Path

PROJECT_ROOT_DIR = Path.cwd().parent
print(PROJECT_ROOT_DIR)
```

```python
images_path = PROJECT_ROOT_DIR.joinpath("datasets", "licence_plate", "images", "test")
print(images_path)
```

```python
image_names = [f.stem for f in images_path.iterdir()]
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

dir1 = PROJECT_ROOT_DIR / "detection" / "results" / "DETR" / "v_001"
dir2 = PROJECT_ROOT_DIR / "detection" / "results" / "YOLOV8S" / "v_001"

CHUNK_SIZE = 500

ocr_dir = PROJECT_ROOT_DIR / "ocr"
ocr_results_filename = "output.json"

with ocr_dir.joinpath("paddle", ocr_results_filename).open("r") as f:
    paddle_results = json.load(f)

with ocr_dir.joinpath("tr_ocr", ocr_results_filename).open("r") as f:
    tr_ocr_results = json.load(f)

for i in range(0, len(image_names), CHUNK_SIZE):
    chunk = image_names[i:i + CHUNK_SIZE]
    num_images = len(chunk)

    fig, axes = plt.subplots(nrows=num_images, ncols=2, figsize=(12, 5 * num_images))
    if num_images == 1:
        axes = [axes]

    axes[0][0].set_title(f"{dir1.parent.name} + TR_ORC", fontsize=14, pad=15)
    axes[0][1].set_title(f"{dir2.parent.name} + PaddleOCR", fontsize=14, pad=15)

    for j, img_name in enumerate(chunk):
        path1 = dir1 / img_name / "results.png"
        path2 = dir2 / img_name / "results.png"
        axes[0][0].set_title(f"{dir1.parent.name} + TR_ORC")
        axes[0][1].set_title(f"{dir2.parent.name} + PaddleOCR")

        try:
            axes[j][0].imshow(mpimg.imread(path1))
        except FileNotFoundError:
            axes[j][0].text(0.5, 0.5, 'Not Found', ha='center')

        try:
            axes[j][1].imshow(mpimg.imread(path2))
        except FileNotFoundError:
            axes[j][1].text(0.5, 0.5, 'Not Found', ha='center')

        text1 = tr_ocr_results.get(img_name, "No OCR result")
        text2 = paddle_results.get(img_name, "No OCR result")

        # Place text underneath using axis coordinates (y=-0.05 puts it just below the image bounds)
        axes[j][0].text(0.5, -0.05, text1, transform=axes[j][0].transAxes,
                        ha='center', va='top', fontsize=11, wrap=True)

        axes[j][1].text(0.5, -0.05, text2, transform=axes[j][1].transAxes,
                        ha='center', va='top', fontsize=11, wrap=True)

        axes[j][0].axis('off')
        axes[j][1].axis('off')

    plt.tight_layout()
    plt.savefig('plot.png', bbox_inches='tight')
    plt.close(fig)
    # plt.show()
```
